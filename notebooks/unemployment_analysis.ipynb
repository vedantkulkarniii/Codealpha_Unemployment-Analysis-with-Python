## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')

# Set visualization style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## 2. Load Dataset

```python
# Load the unemployment dataset
df = pd.read_csv('../data/Unemployment_Rate_upto_11_2020.csv')
print("Dataset loaded successfully!")
print(f"Shape: {df.shape}")
```

## 4. Data Types Inspection

```python
# Check data types of each column
print("Data types of columns:")
print(df.dtypes)
print("\nDetailed info:")
df.info()
```

## 5. Missing Values Analysis

```python
# Check for missing values
print("Missing values per column:")
missing_values = df.isnull().sum()
print(missing_values)

print("\nMissing values percentage:")
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({
    'Missing Count': missing_values,
    'Percentage': missing_percentage
})
print(missing_df)
```

## 6. Duplicate Records Check

```python
# Check for duplicate records
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate records: {duplicate_count}")

if duplicate_count > 0:
    print("\nDuplicate rows:")
    print(df[df.duplicated()])
else:
    print("No duplicate records found.")
```

## 7. Unique Regions/States Analysis

```python
# Inspect unique regions/states
unique_regions = df['Region'].unique()
print(f"Number of unique regions: {len(unique_regions)}")
print("\nList of regions:")
for region in sorted(unique_regions):
    print(f"  - {region}")

# Check area types
print("\nUnique area types:")
print(df['Area'].unique())
print(f"\nArea type distribution:")
print(df['Area'].value_counts())
```

## 8. Date Range Coverage

```python
# Inspect date range
print("Date column sample values:")
print(df['Date'].head(10))

# Get unique dates
unique_dates = df['Date'].unique()
print(f"\nNumber of unique dates: {len(unique_dates)}")

print(f"\nDate range:")
print(f"First date: {df['Date'].iloc[0]}")
print(f"Last date: {df['Date'].iloc[-1]}")

print(f"\nSample of unique dates:")
for date in sorted(unique_dates)[:10]:
    print(f"  - {date}")
```

## 10. Dataset Column Documentation

### Column Descriptions:

1. **Region**: Name of the Indian state or union territory
2. **Date**: Date of observation (DD-MM-YYYY format)
3. **Frequency**: Frequency of data collection (Monthly)
4. **Estimated Unemployment Rate (%)**: Percentage of unemployed individuals in the labor force
5. **Estimated Employed**: Estimated number of employed individuals
6. **Estimated Labour Participation Rate (%)**: Percentage of working-age population participating in the labor force
7. **Area**: Type of area - Rural or Urban

### Key Observations:
- Dataset covers unemployment data from January 2019 to November 2020
- Data is collected monthly for different Indian states
- Includes both rural and urban unemployment data
- Critical period: March 2020 onwards (COVID-19 pandemic period)
- Unemployment rates show significant variation across regions and time periods
