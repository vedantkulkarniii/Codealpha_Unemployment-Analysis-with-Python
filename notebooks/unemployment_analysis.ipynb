## 12. Standardize Column Names

```python
# Create a copy for cleaning
df_clean = df.copy()

# Display original column names
print("Original column names:")
for col in df_clean.columns:
    print(f"  - {col}")

# Standardize column names: lowercase with underscores
df_clean.columns = df_clean.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('(%)', 'pct').str.replace('(', '').str.replace(')', '')

print("\nStandardized column names:")
for col in df_clean.columns:
    print(f"  - {col}")

print(f"\nDataset shape after standardization: {df_clean.shape}")
```

## 13. Clean Whitespace in Categorical Columns

```python
# Identify categorical columns
categorical_cols = df_clean.select_dtypes(include=['object']).columns.tolist()
print(f"Categorical columns: {categorical_cols}")

# Check for whitespace before cleaning
print("\nSample values before cleaning:")
for col in categorical_cols:
    print(f"\n{col}:")
    print(df_clean[col].head(3).tolist())

# Remove leading and trailing whitespace
for col in categorical_cols:
    df_clean[col] = df_clean[col].str.strip()

print("\n" + "="*50)
print("Whitespace cleaned from all categorical columns")
print("="*50)
```

## 14. Convert Date to Datetime Format

```python
# Check current date format
print("Current date column type:")
print(df_clean['date'].dtype)
print("\nSample dates:")
print(df_clean['date'].head())

# Convert date column to datetime
df_clean['date'] = pd.to_datetime(df_clean['date'], format='%d-%m-%Y')

print("\nAfter conversion:")
print(f"Date column type: {df_clean['date'].dtype}")
print("\nSample dates:")
print(df_clean['date'].head())

print("\nDate range:")
print(f"Earliest date: {df_clean['date'].min()}")
print(f"Latest date: {df_clean['date'].max()}")
print(f"Date range span: {(df_clean['date'].max() - df_clean['date'].min()).days} days")
```
